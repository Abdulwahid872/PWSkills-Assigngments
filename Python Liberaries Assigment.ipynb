```python
import pandas as pd
```

```python
df = pd.read_csv('retail_dataset.csv')
```

```python
df.isnull().sum()
```

```python
df.isnull()
```

```python
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].mean())
```

```python
df['Quantity']
```

```python
df['Price'] = df['Price'].fillna(df['Price'].mean())
```

```python
df['Price']
```

```python
df[['Product Category','Region']]
```

```python
df = df.dropna (subset=['Product Category', 'Region'])
```

```python
df.shape
```

```python
import numpy as np
```

```python
df['Revenue'] = np.multiply(df['Quantity'], df['Price'])
```

```python
df.head()
```

```python
Total_Revenue = np.sum(df['Revenue'])
print(Total_Revenue)
```

```python
category_revenue = df.groupby('Product Category')['Revenue'].sum().sort_values(ascending=False)
print(category_revenue)
```

```python
category_revenue = df.groupby('Product Category')['Revenue'].sum().sort_values(ascending=False)
```

```python
top_3 = category_revenue.head(3)
print(top_3)
```

```python
bottom_3 = category_revenue.tail(3)
print(bottom_3)
```

```python
region_revenue = df.groupby('Region')['Revenue'].sum().sort_values(ascending=False)
print(region_revenue)
```

```python
highest_region = region_revenue.idxmax()
highest_value = region_revenue.max()
print("Highest Revenue Region:", highest_region, "-", highest_value)
```

```python
lowest_region = region_revenue.idxmin()
lowest_value = region_revenue.min()
print("Lowest Revenue Region:", lowest_region, "-", lowest_value)
```

```python
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)
```

```python
df['Month'] = df['Date'].dt.month
```

```python
monthly_revenue = df.groupby('Month')['Revenue'].sum()
print(monthly_revenue)
```

```python
mean_revenue = np.mean(df['Revenue'])
median_revenue = np.median(df['Revenue'])
std_revenue = np.std(df['Revenue'])

print("Mean Revenue:", mean_revenue)
print("Median Revenue:", median_revenue)
print("Standard Deviation of Revenue:", std_revenue)
```
